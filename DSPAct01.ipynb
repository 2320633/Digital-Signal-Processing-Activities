## Digital Signal Processing Assessment

This notebook computes and sketches:  
- `x[n]` (original sequence)  
- `y1[n] = x[n-2]` (time shift)  
- `y2[n] = x[-n]` (time reversal)  

Each sketch is provided in its own code cell below.

```python
%matplotlib inline
import matplotlib.pyplot as plt

# original sequence x[n] with origin at n=0 (dict-style to show only nonzero indices)
x = {0: 2, 1: -1, 2: 3, 3: 0, 4: 4}


def show(seq, name):
    # prints n and value pairs, sorted by n
    for n in sorted(seq.keys()):
        print(f" n = {n} -> {seq[n]}")
    print()

print("x[n] (original):")
show(x, "x[n]")

# plot
ns = sorted(x.keys())
vals = [x[n] for n in ns]
plt.stem(ns, vals, basefmt=' ')
plt.title('x[n] (original)')
plt.xlabel('n')
plt.grid(True)
plt.xticks(range(min(ns) - 1, max(ns) + 2))
plt.show()
```

```python
# y1[n] = x[n-2] -> produce dict with shifted indices (nonzero-only)
# build from original dict x
n_y1 = {n + 2: val for n, val in x.items()}

print("y1[n] = x[n-2]:")
show(n_y1, "y1")

# plot
ns = sorted(n_y1.keys())
vals = [n_y1[n] for n in ns]
plt.stem(ns, vals, basefmt=' ')
plt.title('y1[n] = x[n-2]')
plt.xlabel('n')
plt.grid(True)
plt.xticks(range(min(ns) - 1, max(ns) + 2))
plt.show()
```

```python
# y2[n] = x[-n] -> produce dict with reversed indices (nonzero-only)
n_y2 = {-n: val for n, val in x.items()}

print("y2[n] = x[-n]:")
show(n_y2, "y2")

# plot
ns = sorted(n_y2.keys())
vals = [n_y2[n] for n in ns]
plt.stem(ns, vals, basefmt=' ')
plt.title('y2[n] = x[-n]')
plt.xlabel('n')
plt.grid(True)
plt.xticks(range(min(ns) - 1, max(ns) + 2))
plt.show()
```

```python
# Energy of x[n]
E = sum(abs(val) ** 2 for val in x.values())
print(f"Total energy E = Σ |x[n]|^2 = {E}")
```
