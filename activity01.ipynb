# Discrete-Time Signal Processing: Analysis of $x[n] = 2^n$

## 1. Overview
This document provides a technical discussion on generating, analyzing, and visualizing a **real exponential sequence** $x[n] = a^n$ (where $a = 2$) in Discrete-Time Signal Processing (DSP).

---

## 2. Python Implementation Code

The script below uses `numpy` to define the discrete sequence indices and values, and `matplotlib.pyplot` to visualize the signal using a discrete **stem plot**.

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5,10)
x = (2.0 ** n)
print("n =", n)
print("x =", x)

```

```python
plt.stem(n, x)
plt.title('Sequence x[n] = 2^n')
plt.xlabel('n (sample index)')
plt.ylabel('x[n]')
plt.grid(True)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.show()
```
